### Вариант задания 9


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Описание задачи:
Создать базовый класс Order в C#, который будет представлять информацию о
заказах товаров или услуг. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Существующие классы `Item`, `Order`, `OnlineOrder`, `PhysicalOrder` и `SpecializedOrder` расширены конструкторами, свойствами с геттерами и сеттерами, статическими членами и взаимодействием объектов.
Требования к базовому классу Order:
• Атрибуты: ID заказа (OrderId), Дата создания (CreationDate), Сумма заказа
(TotalAmount).
• Методы:
o CalculateTotal(): метод для расчета общей суммы заказа.
o AddItem(Item item): метод для добавления элемента в заказ.
o RemoveItem(Item item): метод для удаления элемента из заказа.

Требования к производным классам:
1. ОнлайнЗаказ (OnlineOrder): Должен содержать дополнительные атрибуты,
такие как Email клиента (CustomerEmail). Метод AddItem() должен быть
переопределен для добавления информации о способе доставки при
добавлении элемента.
2. ФизическийЗаказ (PhysicalOrder): Должен содержать дополнительные
атрибуты, такие как Адрес доставки (DeliveryAddress).
Метод RemoveItem() должен быть переопределен для добавления
информации о возврате товара при удалении элемента.
3. СпециализированныйЗаказ (SpecializedOrder) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Специальные
условия (SpecialConditions). Метод CalculateTotal() должен быть
переопределен для учета специальных условий при расчете общей суммы
заказа.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

// Товар
class Item
{
    public string Name { get; set; }
    public decimal Price { get; set; }
    public int Quantity { get; set; }
    public string Category { get; set; }

    public Item(string name, decimal price, int quantity, string category)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
    }

    public decimal GetTotal()
    {
        return Price * Quantity;
    }

    public void ChangeQuantity(int quantity)
    {
        if (quantity >= 0)
            Quantity = quantity;
    }

    public string GetItemInfo()
    {
        return $"{Name}, категория: {Category}, цена: {Price}, количество: {Quantity}";
    }
}

// Интерфейсы для множественного наследования
interface ITrackable
{
    void TrackOrder();
}

interface INotifiable
{
    void SendNotification();
}

// Базовый класс заказа
class Order
{
    public int OrderId { get; set; }
    public DateTime CreationDate { get; set; }
    public decimal TotalAmount { get; set; }

    public string CustomerName { get; set; }
    public string Status { get; set; }
    public string PaymentMethod { get; set; }

    protected List<Item> Items = new List<Item>();

    public Order(int orderId, string customerName, string paymentMethod)
    {
        OrderId = orderId;
        CustomerName = customerName;
        PaymentMethod = paymentMethod;
        CreationDate = DateTime.Now;
        Status = "Создан";
    }

    public virtual decimal CalculateTotal()
    {
        TotalAmount = 0;

        foreach (Item item in Items)
            TotalAmount += item.GetTotal();

        return TotalAmount;
    }

    public virtual void AddItem(Item item)
    {
        Items.Add(item);
        CalculateTotal();
    }

    public void RemoveItem(Item item)
    {
        Items.Remove(item);
        CalculateTotal();
    }

    public void ConfirmOrder()
    {
        Status = "Подтверждён";
        Console.WriteLine($"Заказ №{OrderId} подтверждён.");
    }

    public void CancelOrder()
    {
        Status = "Отменён";
        Console.WriteLine($"Заказ №{OrderId} отменён.");
    }

    public string GetOrderInfo()
    {
        return $"Заказ №{OrderId}, клиент: {CustomerName}, " +
               $"статус: {Status}, сумма: {CalculateTotal()} руб.";
    }
}

// Простое наследование: OnlineOrder наследуется от Order
class OnlineOrder : Order, ITrackable, INotifiable
{
    public string CustomerEmail { get; set; }
    public string TrackingNumber { get; set; }
    public string DeliveryService { get; set; }
    public bool IsPaid { get; set; }

    public OnlineOrder(int id, string customer, string paymentMethod,
                       string email, string trackingNumber)
        : base(id, customer, paymentMethod)
    {
        CustomerEmail = email;
        TrackingNumber = trackingNumber;
        DeliveryService = "СДЭК";
        IsPaid = false;
    }

    public void PayOnline()
    {
        IsPaid = true;
        Console.WriteLine($"Заказ №{OrderId} оплачен онлайн.");
    }

    public void ChangeDeliveryService(string service)
    {
        DeliveryService = service;
    }

    public void TrackOrder()
    {
        Console.WriteLine(
            $"Заказ №{OrderId}, номер отслеживания: {TrackingNumber}");
    }

    public void SendNotification()
    {
        Console.WriteLine(
            $"Уведомление отправлено на {CustomerEmail}.");
    }

    public override void AddItem(Item item)
    {
        base.AddItem(item);
        Console.WriteLine($"Товар {item.Name} добавлен в онлайн-заказ.");
    }
}

// Физический заказ
class PhysicalOrder : Order
{
    public string DeliveryAddress { get; set; }
    public string CourierName { get; set; }
    public DateTime DeliveryDate { get; set; }
    public bool IsDelivered { get; set; }

    public PhysicalOrder(int id, string customer, string paymentMethod,
                         string address)
        : base(id, customer, paymentMethod)
    {
        DeliveryAddress = address;
        CourierName = "Не назначен";
        DeliveryDate = DateTime.Now.AddDays(2);
        IsDelivered = false;
    }

    public void AssignCourier(string courier)
    {
        CourierName = courier;
        Console.WriteLine($"Курьер {CourierName} назначен.");
    }

    public void DeliverOrder()
    {
        IsDelivered = true;
        Status = "Доставлен";
        Console.WriteLine($"Заказ №{OrderId} доставлен.");
    }

    public string GetDeliveryInfo()
    {
        return $"Адрес: {DeliveryAddress}, курьер: {CourierName}, " +
               $"дата доставки: {DeliveryDate:d}";
    }

    public void ChangeAddress(string address)
    {
        DeliveryAddress = address;
    }
}

// Многоуровневое наследование:
// SpecializedOrder -> OnlineOrder -> Order
class SpecializedOrder : OnlineOrder
{
    public decimal DiscountPercent { get; set; }
    public string PromoCode { get; set; }
    public int BonusPoints { get; set; }
    public string SpecialConditions { get; set; }

    public SpecializedOrder(int id, string customer, string paymentMethod,
                            string email, string trackingNumber,
                            decimal discount, string promoCode)
        : base(id, customer, paymentMethod, email, trackingNumber)
    {
        DiscountPercent = discount;
        PromoCode = promoCode;
        BonusPoints = 0;
        SpecialConditions = "Без дополнительных условий";
    }

    public override decimal CalculateTotal()
    {
        decimal total = base.CalculateTotal();
        TotalAmount = total - total * DiscountPercent / 100;
        return TotalAmount;
    }

    public void AddBonusPoints(int points)
    {
        if (points > 0)
            BonusPoints += points;
    }

    public void ChangePromoCode(string code)
    {
        PromoCode = code;
    }

    public void SetSpecialConditions(string conditions)
    {
        SpecialConditions = conditions;
    }
}

// Демонстрация работы классов
Order order1 = new Order(1, "Иван", "Карта");
order1.AddItem(new Item("Клавиатура", 3000, 1, "Аксессуары"));
order1.AddItem(new Item("Мышь", 1500, 2, "Аксессуары"));
order1.ConfirmOrder();

Console.WriteLine("\n--- Обычный заказ ---");
Console.WriteLine(order1.GetOrderInfo());

OnlineOrder order2 = new OnlineOrder(
    2, "Алексей", "Онлайн", "alex@mail.ru", "TR12345");

order2.AddItem(new Item("Наушники", 5000, 1, "Аудио"));
order2.PayOnline();
order2.TrackOrder();
order2.SendNotification();

Console.WriteLine("\n--- Онлайн-заказ ---");
Console.WriteLine(order2.GetOrderInfo());

PhysicalOrder order3 = new PhysicalOrder(
    3, "Мария", "Наличными", "ул. Ленина, 10");

order3.AddItem(new Item("Монитор", 20000, 1, "Техника"));
order3.AssignCourier("Дмитрий");
Console.WriteLine(order3.GetDeliveryInfo());
order3.DeliverOrder();

Console.WriteLine("\n--- Заказ со скидкой ---");
SpecializedOrder order4 = new SpecializedOrder(
    4, "Ольга", "Онлайн", "olga@mail.ru",
    "TR98765", 10, "SALE10");

order4.AddItem(new Item("Ноутбук", 80000, 1, "Техника"));
order4.AddBonusPoints(100);
order4.SetSpecialConditions("Бесплатная доставка");
order4.PayOnline();
order4.TrackOrder();

Console.WriteLine(order4.GetOrderInfo());
Console.WriteLine($"Промокод: {order4.PromoCode}");
Console.WriteLine($"Бонусные баллы: {order4.BonusPoints}");
Console.WriteLine($"Условия: {order4.SpecialConditions}");
```

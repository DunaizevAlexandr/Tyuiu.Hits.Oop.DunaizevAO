### Вариант задания 9

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Описание задачи:
Создать базовый класс Order в C#, который будет представлять информацию о
заказах товаров или услуг. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Существующие классы `Item`, `Order`, `OnlineOrder`, `PhysicalOrder` и `SpecializedOrder` расширены конструкторами, свойствами с геттерами и сеттерами, статическими членами и взаимодействием объектов.
Требования к базовому классу Order:
• Атрибуты: ID заказа (OrderId), Дата создания (CreationDate), Сумма заказа
(TotalAmount).
• Методы:
o CalculateTotal(): метод для расчета общей суммы заказа.
o AddItem(Item item): метод для добавления элемента в заказ.
o RemoveItem(Item item): метод для удаления элемента из заказа.

Требования к производным классам:
1. ОнлайнЗаказ (OnlineOrder): Должен содержать дополнительные атрибуты,
такие как Email клиента (CustomerEmail). Метод AddItem() должен быть
переопределен для добавления информации о способе доставки при
добавлении элемента.
2. ФизическийЗаказ (PhysicalOrder): Должен содержать дополнительные
атрибуты, такие как Адрес доставки (DeliveryAddress).
Метод RemoveItem() должен быть переопределен для добавления
информации о возврате товара при удалении элемента.
3. СпециализированныйЗаказ (SpecializedOrder) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Специальные
условия (SpecialConditions). Метод CalculateTotal() должен быть
переопределен для учета специальных условий при расчете общей суммы
заказа.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

// Товар
class Item
{
    private string _name;
    private decimal _price;
    private int _quantity;

    // Свойства с геттером и сеттером
    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public decimal Price
    {
        get { return _price; }
        set
        {
            if (value >= 0)
                _price = value;
            else
                throw new ArgumentOutOfRangeException("Цена не может быть отрицательной!");
        }
    }

    public int Quantity
    {
        get { return _quantity; }
        set
        {
            if (value > 0)
                _quantity = value;
            else
                throw new ArgumentOutOfRangeException("Количество должно быть больше нуля!");
        }
    }

    // Конструктор
    public Item(string name, decimal price, int quantity)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
    }

    // Метод расчета стоимости товара
    public decimal GetTotal()
    {
        return Price * Quantity;
    }
}

// Базовый класс заказа
class Order
{
    private int _orderId;
    private DateTime _creationDate;
    private decimal _totalAmount;

    public int OrderId
    {
        get { return _orderId; }
        set
        {
            if (value > 0)
                _orderId = value;
            else
                throw new ArgumentOutOfRangeException("Номер заказа должен быть больше нуля!");
        }
    }

    public DateTime CreationDate
    {
        get { return _creationDate; }
        set { _creationDate = value; }
    }

    public decimal TotalAmount
    {
        get { return _totalAmount; }
        protected set { _totalAmount = value; }
    }

    // Список товаров доступен базовому классу и его наследникам
    protected List<Item> Items = new List<Item>();

    // Статическое свойство: общее количество созданных заказов
    public static int TotalOrders { get; private set; } = 0;

    // Конструктор базового класса
    public Order(int orderId, DateTime creationDate)
    {
        OrderId = orderId;
        CreationDate = creationDate;
        TotalOrders++;
    }

    // Расчет общей суммы заказа
    public virtual void CalculateTotal()
    {
        TotalAmount = 0;

        foreach (Item item in Items)
        {
            // Взаимодействие объектов: Order обращается к объекту Item
            TotalAmount += item.GetTotal();
        }

        Console.WriteLine("Общая сумма заказа: " + TotalAmount + " руб.");
    }

    // Добавление товара
    public virtual void AddItem(Item item)
    {
        Items.Add(item);
        Console.WriteLine("Товар \"" + item.Name + "\" добавлен в заказ №" + OrderId + ".");
    }

    // Удаление товара
    public virtual void RemoveItem(Item item)
    {
        if (Items.Remove(item))
        {
            Console.WriteLine("Товар \"" + item.Name + "\" удален из заказа №" + OrderId + ".");
        }
    }

    // Статический метод
    public static void ShowTotalOrders()
    {
        Console.WriteLine("Всего создано заказов: " + TotalOrders);
    }
}

// Онлайн-заказ
class OnlineOrder : Order
{
    private string _customerEmail;

    public string CustomerEmail
    {
        get { return _customerEmail; }
        set { _customerEmail = value; }
    }

    public OnlineOrder(int orderId, DateTime creationDate, string customerEmail)
        : base(orderId, creationDate)
    {
        CustomerEmail = customerEmail;
    }

    // Переопределение метода AddItem
    public override void AddItem(Item item)
    {
        base.AddItem(item);
        Console.WriteLine("Способ доставки: курьерская доставка.");
    }
}

// Физический заказ
class PhysicalOrder : Order
{
    private string _deliveryAddress;

    public string DeliveryAddress
    {
        get { return _deliveryAddress; }
        set { _deliveryAddress = value; }
    }

    public PhysicalOrder(int orderId, DateTime creationDate, string deliveryAddress)
        : base(orderId, creationDate)
    {
        DeliveryAddress = deliveryAddress;
    }

    // Переопределение метода RemoveItem
    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);
        Console.WriteLine("Товар \"" + item.Name + "\" оформлен на возврат.");
    }
}

// Специализированный заказ
class SpecializedOrder : Order
{
    private string _specialConditions;

    public string SpecialConditions
    {
        get { return _specialConditions; }
        set { _specialConditions = value; }
    }

    public SpecializedOrder(int orderId, DateTime creationDate, string specialConditions)
        : base(orderId, creationDate)
    {
        SpecialConditions = specialConditions;
    }

    // Переопределение расчета суммы с учетом скидки
    public override void CalculateTotal()
    {
        decimal sum = 0;

        foreach (Item item in Items)
        {
            // Взаимодействие SpecializedOrder с объектами Item
            sum += item.GetTotal();
        }

        TotalAmount = sum * 0.9m;

        Console.WriteLine("Специальные условия: " + SpecialConditions);
        Console.WriteLine("Итоговая сумма со скидкой: " + TotalAmount + " руб.");
    }
}

// Демонстрация работы классов и взаимодействия объектов
Item phone = new Item("Телефон", 30000, 1);
Item headphones = new Item("Наушники", 5000, 2);
Item keyboard = new Item("Клавиатура", 4000, 1);

Console.WriteLine("=== ОНЛАЙН-ЗАКАЗ ===");
Order onlineOrder = new OnlineOrder(1, DateTime.Now, "example@mail.ru");
onlineOrder.AddItem(phone);
onlineOrder.AddItem(headphones);
onlineOrder.CalculateTotal();

Console.WriteLine();
Console.WriteLine("=== ФИЗИЧЕСКИЙ ЗАКАЗ ===");
Order physicalOrder = new PhysicalOrder(2, DateTime.Now, "г. Тюмень, ул. Республики, 1");
physicalOrder.AddItem(keyboard);
physicalOrder.AddItem(headphones);
physicalOrder.CalculateTotal();
physicalOrder.RemoveItem(headphones);

Console.WriteLine();
Console.WriteLine("=== СПЕЦИАЛИЗИРОВАННЫЙ ЗАКАЗ ===");
Order specializedOrder = new SpecializedOrder(3, DateTime.Now, "Скидка 10% для постоянного клиента");
specializedOrder.AddItem(phone);
specializedOrder.AddItem(keyboard);
specializedOrder.CalculateTotal();

Console.WriteLine();
Order.ShowTotalOrders();
Console.WriteLine("=== ВЫПОЛНЕНИЕ ЗАВЕРШЕНО ===");
```
